```python
# Define our custom financial calculator function
def calculate_net_profit(gross_revenue, expenses):
    # Calculate the margin formula
    net_profit = gross_revenue - expenses
    profit_margin = (net_profit / gross_revenue) * 100
    
    # Return the calculated values back to the user
    return net_profit, profit_margin

# Let's run (call) our custom machine with some sample data!
revenue_input = 7500
expenses_input = 3200

profit, margin = calculate_net_profit(revenue_input, expenses_input)

print(f"--- Double Eagle Financial Calculator Engine ---")
print(f"Net Profit: ${profit}")
print(f"Profit Margin: {margin:.1f}%")
```

```python
# Create a master blueprint for tracking clients
class ClientAccount:
    # The initializer function sets up the basic profile details
    def __init__(self, name, current_balance):
        self.name = name
        self.balance = current_balance
        
    # A custom action (method) allowing us to process updates
    def receive_payment(self, amount):
        self.balance -= amount
        print(f"Payment received from {self.name}: ${amount}")
        print(f"New outstanding balance: ${self.balance}")

# Let's use our blueprint to create a real individual client object!
client_one = ClientAccount("Double Eagle Client A", 1500)

print(f"--- Client Ledger Initialized ---")
print(f"Client Name: {client_one.name} | Balance: ${client_one.balance}")

# Process an automation payment update action
client_one.receive_payment(400)

```
